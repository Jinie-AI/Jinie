```
from google.colab import files
from pathlib import Path
import zipfile, os
uploaded = files.upload()
archive = next(iter(uploaded))
with zipfile.ZipFile(archive) as z:
    for member in z.infolist():
        target = (Path("/content/jinie") / member.filename).resolve()
        if not target.is_relative_to(Path("/content/jinie")):
            raise ValueError("Unsafe ZIP path")
    z.extractall("/content/jinie")
project = next(Path("/content/jinie").rglob("training/train_code.py" )).parents[1]
os.chdir(project)
print(project)
```
